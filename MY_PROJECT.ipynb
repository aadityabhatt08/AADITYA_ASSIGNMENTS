RETAIL STORE SALES ANALYSIS


```python
#NAME -->AADITYA BHATT
#ADMISSION NO--> 2026B15410108
#CLASS & SEC--> CSE(DS)-1


import pandas as pd


class RetailSalesAnalyzer:
    """A class that stores sales data and provides analysis methods."""

    def __init__(self, dataframe):
        self.df = dataframe.copy()
        self.prepare_data()

    def prepare_data(self):
        """Clean data and calculate sales amount."""
        self.df["Date"] = pd.to_datetime(self.df["Date"], errors="coerce")
        self.df["Quantity"] = pd.to_numeric(self.df["Quantity"], errors="coerce")
        self.df["UnitPrice"] = pd.to_numeric(self.df["UnitPrice"], errors="coerce")

        self.df = self.df.dropna(
            subset=["Date", "Quantity", "UnitPrice", "Product", "Category", "Region"]
        )

        self.df["Sales"] = self.df["Quantity"] * self.df["UnitPrice"]
        self.df["Month"] = self.df["Date"].dt.to_period("M").astype(str)

    def show_basic_info(self):
        print("\n--- DATA OVERVIEW ---")
        print("Number of records:", len(self.df))
        print("Number of products:", self.df["Product"].nunique())
        print("Categories:", ", ".join(sorted(self.df["Category"].unique())))
        print("\nFirst five records:")
        print(self.df.head().to_string(index=False))

    def total_sales(self):
        return self.df["Sales"].sum()

    def sales_by_category(self):
        return self.df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

    def sales_by_region(self):
        return self.df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

    def top_products(self, count=5):
        return (
            self.df.groupby("Product")["Sales"]
            .sum()
            .sort_values(ascending=False)
            .head(count)
        )

    def monthly_sales(self):
        return self.df.groupby("Month")["Sales"].sum().sort_index()

    def best_selling_products_by_quantity(self, count=5):
        return (
            self.df.groupby("Product")["Quantity"]
            .sum()
            .sort_values(ascending=False)
            .head(count)
        )

    def save_summary(self):
        self.sales_by_category().to_csv("sales_by_category.csv", header=["Sales"])
        self.sales_by_region().to_csv("sales_by_region.csv", header=["Sales"])
        self.top_products().to_csv("top_products.csv", header=["Sales"])
        self.monthly_sales().to_csv("monthly_sales.csv", header=["Sales"])

        print("\nSummary files saved in the project folder.")


def load_data():
    """Load data from the existing CSV file."""

    dataframe = pd.read_csv("retail_store_sales_sample.csv")
    print("Loaded data from retail_store_sales_sample.csv")

    required_columns = {
        "OrderID", "Date", "Product", "Category",
        "Quantity", "UnitPrice", "Region"
    }

    missing = required_columns - set(dataframe.columns)

    if missing:
        raise ValueError(
            "Missing required CSV columns: " + ", ".join(sorted(missing))
        )

    return dataframe


def display_report(analyzer):
    print("\n========== RETAIL SALES ANALYSIS ==========")

    analyzer.show_basic_info()

    print("\n--- KEY PERFORMANCE INDICATORS ---")
    print(f"Total revenue: Rs. {analyzer.total_sales():,.2f}")
    print(f"Total units sold: {analyzer.df['Quantity'].sum():,.0f}")
    print(f"Average order-line sales: Rs. {analyzer.df['Sales'].mean():,.2f}")

    print("\n--- SALES BY CATEGORY ---")
    print(analyzer.sales_by_category().to_string())

    print("\n--- SALES BY REGION ---")
    print(analyzer.sales_by_region().to_string())

    print("\n--- TOP PRODUCTS BY REVENUE ---")
    print(analyzer.top_products().to_string())

    print("\n--- BEST-SELLING PRODUCTS BY QUANTITY ---")
    print(analyzer.best_selling_products_by_quantity().to_string())

    print("\n--- MONTHLY SALES ---")
    print(analyzer.monthly_sales().to_string())

    best_category = analyzer.sales_by_category().idxmax()
    best_region = analyzer.sales_by_region().idxmax()
    best_product = analyzer.top_products(1).index[0]

    print("\n--- BUSINESS INSIGHTS ---")
    print(f"Highest-revenue category: {best_category}")
    print(f"Highest-revenue region: {best_region}")
    print(f"Top product by revenue: {best_product}")


def main():
    try:
        data = load_data()
        analyzer = RetailSalesAnalyzer(data)
        display_report(analyzer)
        analyzer.save_summary()

    except (ValueError, KeyError, FileNotFoundError) as error:
        print("\nCould not complete the analysis:", error)


if __name__ == "__main__":
    main()
```
